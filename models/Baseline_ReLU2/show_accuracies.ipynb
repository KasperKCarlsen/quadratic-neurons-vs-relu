```python
import pickle
import matplotlib.pyplot as plt

# Load accuracies.pkl
with open('accuracies.pkl', 'rb') as f:
    accuracies = pickle.load(f)

# Put into lists, assuming dictionary with epochs as keys
epochs = list(accuracies.keys())
acc_values = list(accuracies.values())

# Plot the accuracies
plt.figure(figsize=(10, 6))
plt.plot(epochs, acc_values, marker='o')
plt.title('Validation Accuracies over Epochs')
plt.xlabel('Epoch')
plt.ylabel('Accuracy (%)')
plt.grid(True)
plt.show()
```
